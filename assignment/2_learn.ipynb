### ชนิดข้อมูล (Data Types)

Python มีชนิดข้อมูลพื้นฐานที่ใช้ในบทเรียนนี้ ดังนี้

| ชนิดข้อมูล | ความหมาย | ตัวอย่าง |
|---|---|---|
| `int` | จำนวนเต็ม | `90` |
| `float` | จำนวนทศนิยม | `45.5` |
| `str` | ข้อความ ต้องใส่เครื่องหมายคำพูด | `"elbow"` |
| `bool` | ค่าทางตรรกะ จริงหรือเท็จ | `True` / `False` |

```python
robot_name = "Robo"      # str
shoulder = 90            # int
elbow = 45.5             # float
is_active = True         # bool
print(type(robot_name), type(shoulder), type(elbow), type(is_active))
```

**ลองทายผลลัพธ์ก่อนกด RUN:** `type("90")` จะได้ชนิดข้อมูลเป็น `int` หรือ `str`

**คำตอบของผู้เรียน:** (พิมพ์คำตอบตรงนี้ แล้วรันโค้ดด้านบนเพื่อตรวจสอบ)

## 2. List = รายการเก็บข้อมูลหลายค่า

List ใช้จัดเก็บข้อมูลหลายค่าไว้ในตัวแปรเดียว โดยข้อมูลแต่ละตำแหน่งมีหมายเลข Index กำกับ ซึ่งเริ่มนับจาก 0

ตัวอย่างการเก็บค่ามุมของข้อต่อหุ่นยนต์

- Index 0 คือ `shoulder`
- Index 1 คือ `elbow`
- Index 2 คือ `wrist`

สามารถเลือกดูหรือเปลี่ยนค่าของแต่ละตำแหน่งได้ด้วย Index

```python
joint_angles = [0, 90, 45]
print(joint_angles[1])        # อ่านค่าที่ Index 1
joint_angles[1] = 120         # เปลี่ยนค่าที่ Index 1
print(joint_angles)
print(len(joint_angles))      # นับจำนวนข้อมูลใน List

try:
    print(joint_angles[3])    # ตำแหน่งที่ไม่มีอยู่จริง
except IndexError as err:
    print("เกิด IndexError:", err)
```

**ลองทายผลลัพธ์ก่อนกด RUN:** คำสั่ง `print(joint_angles[1])` ในครั้งแรกจะแสดงค่าใด และเมื่อเรียก `joint_angles[3]` จะเกิดอะไรขึ้น

**คำตอบของผู้เรียน:** (พิมพ์คำตอบตรงนี้ แล้วรันโค้ดด้านบนเพื่อตรวจสอบ)

## 3. Condition = การตรวจสอบเงื่อนไข

คำสั่ง `if`, `elif` และ `else` ใช้ตรวจสอบเงื่อนไขและกำหนดให้โปรแกรมทำงานแตกต่างกันตามผลที่ตรวจสอบ

ในการควบคุมหุ่นยนต์ สามารถนำมาใช้ตรวจสอบค่ามุมของข้อต่อ เพื่อไม่ให้หมุนเกินขอบเขตที่กำหนดไว้ คือ 0–180 องศา

**ข้อควรจำ**
- บรรทัดคำสั่งภายใต้ `if`, `elif` และ `else` ต้องย่อหน้าให้ถูกต้อง
- ต้องใส่เครื่องหมาย `:` ต่อท้ายคำสั่ง `if`, `elif` และ `else`

```python
angle = 215
if angle > 180:
    angle = 180
    print("เกินขอบเขต ปรับเป็น 180")
elif angle < 0:
    angle = 0
    print("ต่ำกว่า 0 ปรับเป็น 0")
else:
    print("มุมอยู่ในช่วงปลอดภัย")
print("มุมสุดท้าย:", angle)
```

**ลองทายผลลัพธ์ก่อนกด RUN:** หากเปลี่ยนค่าเป็น `angle = 50` โปรแกรมจะแสดงข้อความใด และมุมสุดท้ายจะมีค่าเท่าใด

**คำตอบของผู้เรียน:** (พิมพ์คำตอบตรงนี้ แล้วรันโค้ดด้านบนเพื่อตรวจสอบ)

## 4. Loop = การทำซ้ำ

คำสั่ง `for` ใช้สั่งให้โปรแกรมทำงานซ้ำกับข้อมูลใน List ช่วยลดการเขียนคำสั่งซ้ำทีละข้อต่อ

การใช้งานในบทเรียนนี้แบ่งเป็น 2 รูปแบบ

- `for a in joint_angles:` ใช้เมื่อต้องการอ่านค่าของแต่ละรายการ
- `for i in range(len(joint_angles)):` ใช้เมื่อต้องการเข้าถึงตำแหน่งของข้อมูลและแก้ไขค่าใน List

```python
joint_angles = [0, 90, 45]
for a in joint_angles:
    print("มุมนี้คือ", a)

for i in range(len(joint_angles)):
    joint_angles[i] = joint_angles[i] + 10
print(joint_angles)
```

**ลองทายผลลัพธ์ก่อนกด RUN:** คำสั่ง `print(joint_angles)` บรรทัดสุดท้ายจะแสดงค่าใด หลังจากโปรแกรมทำงานครบทั้งสอง Loop

**คำตอบของผู้เรียน:** (พิมพ์คำตอบตรงนี้ แล้วรันโค้ดด้านบนเพื่อตรวจสอบ)

## สรุปเนื้อหาบทเรียน

| แนวคิด | ตัวอย่างโค้ด | การนำไปใช้กับหุ่นยนต์ |
|---|---|---|
| Variable | `shoulder = 90` | จัดเก็บค่ามุมของข้อต่อ |
| List | `joint_angles = [0, 90, 45]` | จัดเก็บค่ามุมของข้อต่อหลายตำแหน่ง |
| Index | `joint_angles[1]` | เข้าถึงข้อมูลตามตำแหน่งใน List |
| Condition | `if angle > 180:` | ตรวจสอบและจำกัดค่ามุมให้อยู่ในช่วงที่กำหนด |
| Loop | `for i in range(len(joint_angles)):` | ทำซ้ำเพื่อปรับค่ามุมของข้อต่อทุกตำแหน่ง |
