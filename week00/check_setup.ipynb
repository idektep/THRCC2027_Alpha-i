# 🧪 Week 0: ตรวจว่าเครื่องพร้อมแล้วหรือยัง
⏱️ 2 นาที

**ทำตามนี้:**
1. มุมขวาบนกด **Select Kernel** → Python Environments → เลือก **.venv**
2. กดรันเซลล์ด้านล่าง (หรือกด **Run All**)
3. อ่านผล: ✅ ผ่าน · ⚠️ เตือน (เรียนต่อได้) · ❌ ต้องแก้ (มีวิธีแก้ใต้บรรทัดนั้น)

```python
# 🔎 ตัวตรวจสภาพแวดล้อม: แค่กดรันเซลล์นี้ ไม่ต้องแก้อะไร
import sys, platform, importlib
from pathlib import Path

ICON = {"ok": "✅", "warn": "⚠️", "fail": "❌"}
results = []  # (ชื่อ, สถานะ, วิธีแก้)

def add(name, status, fix=""):
    results.append((name, status, fix))

# 0) ระบบปฏิบัติการ
os_name = {"Darwin": "macOS"}.get(platform.system(), platform.system())
add(f"ระบบปฏิบัติการ: {os_name}", "ok")

# 1) เวอร์ชัน Python (ต้อง 3.10 ขึ้นไป แนะนำ 3.13)
v = sys.version_info
ver = f"{v.major}.{v.minor}.{v.micro}"
if (v.major, v.minor) < (3, 10):
    add(f"Python {ver} (เก่าเกินไป ต้องเป็น 3.10 ขึ้นไป)", "fail",
        "ติดตั้ง Python 3.13 ตามสไลด์ Week 0 ขั้นที่ 1 แล้วสร้าง .venv ใหม่")
elif (v.major, v.minor) != (3, 13):
    add(f"Python {ver} (ชุดแข่งขันแนะนำ 3.13)", "warn",
        "เรียนสัปดาห์ 1 ต่อได้ แต่ควรติดตั้ง Python 3.13 ตามสไลด์ Week 0 แล้วสร้าง .venv ใหม่ก่อนถึงชุดแข่งขัน")
else:
    add(f"Python {ver}", "ok")

# 2) ใช้ Virtual Environment (.venv) อยู่ไหม
if sys.prefix != getattr(sys, "base_prefix", sys.prefix):
    add("ใช้ Virtual Environment (.venv) อยู่", "ok")
else:
    add("ยังไม่ได้ใช้ .venv", "warn",
        "มุมขวาบนของ Notebook กด Select Kernel → Python Environments → เลือก .venv แล้วรันเซลล์นี้ใหม่")

# 3) library ที่คอร์สต้องใช้
for lib in ["numpy", "matplotlib", "ipykernel", "ipywidgets"]:
    try:
        m = importlib.import_module(lib)
        add(f"{lib} {getattr(m, '__version__', '')}".strip(), "ok")
    except Exception:
        add(f"{lib} ยังไม่ได้ติดตั้ง", "fail",
            f"เปิด Terminal ที่มี (.venv) หน้าบรรทัด แล้วรัน: pip install -r requirements.txt  "
            f"(หรือรัน: {sys.executable} -m pip install {lib})  แล้วเลือก Kernel เป็น .venv")

# 4) เปิดโฟลเดอร์คอร์สถูกต้องไหม
root = next((p for p in [Path.cwd(), *Path.cwd().parents] if (p / "robot_course").exists()), None)
if root:
    add(f"พบโฟลเดอร์คอร์ส ({root.name})", "ok")
    if (root / "week01" / "assignment" / "1_mission.ipynb").exists():
        add("พบไฟล์สัปดาห์ 1", "ok")
    else:
        add("ไม่พบไฟล์สัปดาห์ 1", "warn", "ดาวน์โหลดหรือ git clone คอร์สจาก GitHub ให้ครบ แล้วเช็กว่ามีโฟลเดอร์ week01")
else:
    add("ไม่พบโฟลเดอร์ robot_course", "fail",
        "ใน VS Code เลือก File → Open Folder แล้วเปิดทั้งโฟลเดอร์ robot-rl-course (ไม่ใช่เปิดไฟล์เดียว) แล้วเปิดไฟล์นี้ใหม่")

# แสดงผลการตรวจ
for name, status, fix in results:
    print(ICON[status], name)
    if fix and status != "ok":
        print("    ↳ วิธีแก้:", fix)

# 5) ทดสอบวาดหุ่นยนต์ตัวแรก (ทำเมื่อผ่านข้อก่อนหน้าแล้วเท่านั้น)
if root and not any(s == "fail" for _, s, _ in results):
    try:
        if str(root) not in sys.path:
            sys.path.insert(0, str(root))
        from robot_course import show_robot
        print("\n🤖 หุ่นยนต์ตัวแรกของเธอ:")
        show_robot({"shoulder": 90, "elbow": 45, "wrist": 0})
        print("✅ วาดภาพหุ่นยนต์ได้")
    except Exception as e:
        results.append(("วาดภาพหุ่นยนต์ไม่สำเร็จ", "fail", str(e)))
        print("❌ วาดภาพหุ่นยนต์ไม่สำเร็จ")
        print("    ↳ รายละเอียด:", e)
        print("    ↳ วิธีแก้: ติดตั้ง matplotlib และ ipywidgets แล้วรีสตาร์ท Kernel (ปุ่ม Restart) จากนั้นรันใหม่")

# สรุปผล
fails = [r for r in results if r[1] == "fail"]
warns = [r for r in results if r[1] == "warn"]
print()
if not fails:
    print("🤖 ผ่านทั้งหมด พร้อมเริ่มสัปดาห์ 1")
    if warns:
        print(f"   (มีคำเตือน {len(warns)} ข้อ ดูข้อ ⚠️ ด้านบน แต่เริ่มเรียนได้)")
else:
    print(f"ยังมี {len(fails)} ข้อที่ต้องแก้ (ข้อ ❌) ทำตามวิธีแก้ แล้วรันเซลล์นี้ใหม่ได้ไม่จำกัดครั้ง")
```

เปิดไฟล์ `week01/assignment/1_mission.ipynb` เพื่อเริ่มสัปดาห์ 1 (ใช้ `.venv` ตัวเดิม ไม่ต้องสร้างใหม่)

## ❌ ยังติด? ดูตารางนี้
| อาการ | วิธีแก้ |
|---|---|
| ไม่พบโฟลเดอร์ `robot_course` | File → Open Folder แล้วเปิดทั้งโฟลเดอร์ `robot-rl-course` |
| `ModuleNotFoundError` หรือขึ้น ❌ ที่ library | เลือก Kernel เป็น `.venv` แล้วรัน `pip install -r requirements.txt` ใน Terminal ที่มี `(.venv)` |
| Windows: `python is not recognized` | ติดตั้ง Python ใหม่ แล้วติ๊ก **Add python.exe to PATH** |
| Windows: activate `.venv` ไม่ได้ | รัน `Set-ExecutionPolicy -Scope CurrentUser RemoteSigned` แล้วลองใหม่ |
| Mac: `command not found: python` | ใช้ `python3` แทน `python` |
| ไม่เห็น Kernel `.venv` | ติดตั้ง Extension Python และ Jupyter แล้วเปิดโฟลเดอร์คอร์สใหม่ |
| ติดตั้งไม่ได้เลย | ใช้ Colab สำรอง (ขอลิงก์จากผู้จัด) |
