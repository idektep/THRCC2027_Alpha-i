**เมื่อจบสัปดาห์นี้ คุณจะสามารถ:**
- เก็บมุมของข้อต่อไว้ในตัวแปรและ List

- ป้องกันไม่ให้ข้อต่อหมุนเกินช่วง 0–180 องศา
- ปรับทุกข้อต่อพร้อมกันด้วย Loop

**ลำดับการเรียน:** 1_mission → 2_learn → 3_try_it → 4_challenge → 5_check
