## 2️⃣ List = ตู้ลิ้นชัก
เก็บหลายค่าเรียงกัน แต่ละลิ้นชักมีเลขที่ **นับจาก 0** (ลิ้นชัก 0 = shoulder, 1 = elbow, 2 = wrist)

```python
joint_angles = [0, 90, 45]
print(joint_angles[1])        # อ่านลิ้นชัก 1
joint_angles[1] = 120         # เปลี่ยนค่าในลิ้นชัก 1
print(joint_angles)
print(len(joint_angles))      # นับจำนวนลิ้นชัก

try:
    print(joint_angles[3])    # ลิ้นชักที่ไม่มีอยู่จริง
except IndexError as err:
    print("เจอ IndexError:", err)
```

🤔 **ทายก่อนรัน:** ตอนที่ `print(joint_angles[1])` ครั้งแรก จะพิมพ์อะไร? และ `joint_angles[3]` เกิดอะไรขึ้น?

✏️ **คำตอบของฉัน:** (พิมพ์ทับตรงนี้ แล้วรันโค้ดด้านบนเพื่อเช็ก)

## 3️⃣ Condition = ยามเฝ้าข้อต่อ
`if` คอยกันไม่ให้ข้อต่อหมุนเกิน 180° หรือต่ำกว่า 0°

⚠️ บรรทัดที่อยู่ใต้ `if` / `elif` / `else` ต้อง **ย่อหน้า** และท้ายบรรทัด `if` ต้องมี `:`

```python
angle = 215
if angle > 180:
    angle = 180
    print("เกินแล้ว! ปรับเหลือ 180")
elif angle < 0:
    angle = 0
    print("ต่ำกว่า 0 ปรับเหลือ 0")
else:
    print("มุมปลอดภัย")
print("มุมสุดท้าย:", angle)
```

🤔 **ทายก่อนรัน:** ถ้าเปลี่ยนเป็น `angle = 50` จะพิมพ์อะไร?

✏️ **คำตอบของฉัน:** (พิมพ์ทับตรงนี้ แล้วรันโค้ดด้านบนเพื่อเช็ก)

## 4️⃣ Loop = สั่งรวดเดียว
`for` ทำซ้ำให้ทุกค่าใน List แทนการสั่งทีละข้อต่อ
- `for a in joint_angles:` ใช้เมื่อ **แค่อ่านค่า**
- `for i in range(len(joint_angles)):` ใช้เมื่อต้อง **แก้ค่า** ใน List

```python
joint_angles = [0, 90, 45]
for a in joint_angles:
    print("มุมนี้คือ", a)

for i in range(len(joint_angles)):
    joint_angles[i] = joint_angles[i] + 10
print(joint_angles)
```

🤔 **ทายก่อนรัน:** บรรทัดสุดท้าย `print(joint_angles)` จะพิมพ์อะไร?

✏️ **คำตอบของฉัน:** (พิมพ์ทับตรงนี้ แล้วรันโค้ดด้านบนเพื่อเช็ก)
